## Research Question

How do El Niño, La Niña, and neutral winters differ in wave height, wave period, and surf consistency near Torrey Pines in San Diego from December 2001 through February 2026?

We will compare complete December through February winters using daily wave measurements and NOAA's historical climate classifications. Our main outcomes will be mean daily significant wave height and the percentage of days meeting a selected wave range, initially 0.5 to 1.5 meters with a peak period of 8 to 14 seconds. We will count a day as meeting that range when at least half of its valid measurements are inside both limits, and we will test other ranges to see how much our choices change the results. This is a statistical comparison of historical conditions, and our selected range describes offshore waves rather than proving that a day is safe for beginners.

## Background and Prior Work

We want to know whether El Niño brings more useful surfing days or mainly bigger waves. Bigger waves may appeal to experienced surfers but be less useful for people looking for smaller conditions. NOAA's historical Oceanic Niño Index, or ONI, identifies sustained warm and cold Pacific conditions. We will use historical ERSSTv5 ONI consistently, while noting that NOAA now uses the Relative Oceanic Niño Index for official monitoring.1
Barnard and colleagues found unusually high wave energy along the West Coast during the 2015 to 2016 El Niño. This gives us a reason to expect larger waves, but it does not show whether less experienced surfers had more useful days. We will compare several winters rather than base our conclusion on one strong event.2
Smith and Barnard found that wave direction during that event differed from earlier El Niño events, showing why we should check individual winters.3 We will use CDIP's Torrey Pines buoy measurements to compare wave height, period, and consistency across climate phases.4 Our focus is how often selected conditions occur, while recognizing that offshore measurements do not show the exact waves breaking at a beach.

## Hypothesis


We predict that El Niño winters will have higher average daily wave heights than neutral winters, but fewer days within our selected smaller-wave range. Earlier research found unusually energetic waves during a major El Niño, and larger waves could fall outside our range. We will also compare La Niña winters without assuming they are simply the opposite of El Niño.

## Data

Ideal Data:

We would want 20 to 30 years of regular measurements from several San Diego surf beaches, covering multiple separate climate events. Variables would include date, location, breaking-wave height, period, direction, wind, tides, climate phase, and quality flags. Calibrated buoys and nearshore sensors would collect thousands of observations, stored as one row per location and timestamp with a separate climate table. Local wave observations would help check how well offshore measurements represent actual surf.

Dataset 1: CDIP wave measurements

The free Torrey Pines Outer buoy archive at https://thredds.cdip.ucsd.edu/thredds/catalog/cdip/archive/100p1/catalog.html?dataset=CDIP_Archive/100p1/100p1_historic.nc  has NetCDF downloads and remote access without a paid subscription or application. The archive previously inspected listed about 436,000 observations from 2001 through April 2026, exceeding the course's approximately 1,000-observation requirement. We will use timestamp, significant wave height, peak period, mean period, direction, and quality flags. We still need to verify usable counts after downloading and cleaning, and these offshore measurements will not represent every San Diego beach.


Dataset 2: NOAA climate data

NOAA's historical ONI table at https://www.cpc.ncep.noaa.gov/products/analysis_monitoring/enso/oni/v5/ is free and includes seasonal values from 1950 onward. We will use year, three-month season, temperature anomaly, and historical El Niño, La Niña, or neutral classification. We will follow NOAA's sustained-event rules instead of treating one warm or cold reading as a whole event. This table gives us climate information to join with the waves, but overlapping seasons and repeated labels are not independent climate events.

Cleaning and combining the data
We will use complete December through February winters, subject to coverage. Winter 2002 means December 2001 through February 2002, matched to NOAA's DJF 2002 classification. We will check missing values, duplicate records, quality flags, and timestamps, then convert wave times to local San Diego time.
We will require at least 75% of expected measurements before using a daily summary, after checking the actual sampling schedule. We will report valid days, winters, and distinct events in each phase and test whether excluding transition winters changes the results. Missing measurements will stay missing rather than becoming zeros.

Analysis and figures
1. Compare winter-level average wave heights across climate phases and estimate uncertainty while accounting for related observations and winters within the same event.
2. Use a simple model for proportions to test whether climate-index strength is related to the percentage of days inside our selected wave range. Check alternative ranges, a time trend, and whether leaving out major events changes the result.


Our three main figures will show wave-height distributions by phase, the percentage of days in selected ranges, and a timeline connecting climate values with wave outcomes. We will label units, show sample counts and uncertainty, and explain each figure. Thousands of buoy readings do not mean thousands of independent El Niño events, so we will keep the analysis small and report that limitation.
